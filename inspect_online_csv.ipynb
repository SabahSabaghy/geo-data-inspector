# Geo Data Inspector: Preview an Online CSV Before Saving

A small reusable workflow for inspecting an online CSV with pandas before deciding whether to save it locally.

Set `url`, run the notebook, review the structure and first records, then change `save_csv` to `True` only if you want a local copy.

```python
import pandas as pd

# URL of the source CSV file
url = ""

# Read the CSV directly into a pandas DataFrame
df = pd.read_csv(url)
```

## Inspect the dataset

```python
# Summary: row/column information, column names and data types
df.info()

# Preview the first 10 records
display(df.head(10))
```

## Optional: save locally

Leave `save_csv = False` while exploring. If the data looks useful, change it to `True` and rerun this cell.

```python
save_csv = False
output_path = r"C:\data\my_data.csv"

if save_csv:
    df.to_csv(output_path, index=False)
    print(f"CSV saved successfully: {output_path}")
else:
    print("Preview only. Set save_csv = True to save the CSV locally.")
```
